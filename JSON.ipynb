```python
x = {
    'State': 'AP',
    'city': 'Vijayawada',
    'country': 'India'
}
```

```python
print(type(x))
```

```python
#dumps:python objects into json string
y=json.dumps(x)
print(y)
print(type(y))
```

```python
y=json.dumps(x,indent =50)
print(y)
print(type(y))
```

```python
# loads :json string python object
x=json.loads(y)
print(x,type(x))
```

```python
my_dict={'flower':"rose",
         'fruit':"Mango",
        }
```

```python
#converting python obj into json string using dump
with open("my_dict.json", "w") as f:
    json.dump(my_dict, f)
    print(type(my_dict))
```

```python
# loads:json string into python obj
with open("my_dict.json", "r") as f:
    data = json.load(f)
    print(data)
    print(type(data))
```

```python
#writing the dict value using dump
with open("my_dict.json", "w") as f:
    json.dump({'plant':"Tulasi"},f)
```

```python
with open ("my_dict.json","r") as f:
    data=json.load(f)
    print(data)
    print(type(data))
```
